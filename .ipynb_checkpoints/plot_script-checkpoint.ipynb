```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data 
df = pd.read_csv("spo26_097.csv")

# Remove NO RESPONSE rows
df = df[df["Response"] == "OK"]

# Drop rows with NA values (just in case)
df = df.dropna()

# Convert Time column to datetime
df["DateTime"] = pd.to_datetime(df["Date"] + " " + df["Time"])

# Convert Voltage and Current to numeric
df["Voltage"] = pd.to_numeric(df["Voltage"])
df["Current"] = pd.to_numeric(df["Current"])

# Plot
plt.figure()
plt.plot(df["DateTime"], df["Voltage"], label="Voltage")
plt.plot(df["DateTime"], df["Current"], label="Current")

plt.xlabel("Time")
plt.ylabel("Value")
plt.title("Voltage and Current Over Time")
plt.legend()
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
power = df["Voltage"] * df["Current"]
plt.figure()
plt.plot(df["DateTime"], power, label="Power")
plt.xlabel("Time")
plt.ylabel("Power (W)")
plt.title("Power Over Time")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
temperature = df["Temp"]
plt.figure()
plt.plot(df["DateTime"], temperature, label="Temperature", color="orange")
plt.xlabel("Time")
plt.ylabel("Temperature (°C)")
plt.title("Temperature Over Time")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#truncate the data from 2:30 to 3:30
start_time = pd.to_datetime("2026-04-07 14:05:00")
end_time = pd.to_datetime("2026-04-07 15:40:00")
truncated_df = df[(df["DateTime"] >= start_time) & (df["DateTime"] <= end_time)]
plt.figure()
plt.plot(truncated_df["DateTime"], truncated_df["Voltage"], label="Voltage")
plt.plot(truncated_df["DateTime"], truncated_df["Current"], label="Current")
plt.xlabel("Time")
plt.ylabel("Value")
plt.title("Voltage and Current from 2:30 to 3:30")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#plot the power in that same time range
truncated_power = truncated_df["Voltage"] * truncated_df["Current"]
plt.figure()
plt.plot(truncated_df["DateTime"], truncated_power, label="Power", color="green")
plt.xlabel("Time")
plt.ylabel("Power (W)")
plt.title("Power from 2:30 to 3:30")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
